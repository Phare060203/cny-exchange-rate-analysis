```python
# Cell 1: 安装库（只需运行一次）
%pip install akshare pandas matplotlib seaborn

# Cell 2: 导入库
import akshare as ak
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# 创建figures目录
if not os.path.exists('figures'):
    os.makedirs('figures')

# Cell 3: 获取真实汇率数据（AKShare）
print("正在从中国外汇交易中心获取官方汇率数据...")

# 获取美元兑人民币
usd_df = ak.currency_boc_sina(symbol="美元")
# 获取欧元兑人民币
eur_df = ak.currency_boc_sina(symbol="欧元")

# 获取数据并重命名列
usd_df = ak.currency_boc_sina(symbol="美元")
eur_df = ak.currency_boc_sina(symbol="欧元")

usd_df = usd_df.rename(columns={'日期': 'Date', '中行折算价': 'Close'})
eur_df = eur_df.rename(columns={'日期': 'Date', '中行折算价': 'Close'})

# 日期处理
usd_df['Date'] = pd.to_datetime(usd_df['Date'])
eur_df['Date'] = pd.to_datetime(eur_df['Date'])
usd_df.set_index('Date', inplace=True)
eur_df.set_index('Date', inplace=True)

# 排序
usd_df.sort_index(inplace=True)
eur_df.sort_index(inplace=True)

# 可选：只保留2015年之后的数据
usd_df = usd_df.loc['2015-01-01':]
eur_df = eur_df.loc['2015-01-01':]

# ========== 关键：除以 100（因为原始数据是 100 外币兑人民币） ==========
usd_df['Close'] = usd_df['Close'] / 100
eur_df['Close'] = eur_df['Close'] / 100
# ====================================================================

# 然后赋值给 usdcny / eurcny
usdcny = usd_df[['Close']]
eurcny = eur_df[['Close']]
# Cell 4: 合并数据
combined = pd.DataFrame()
combined['USD/CNY'] = usdcny['Close']
combined['EUR/CNY'] = eurcny['Close']

# ============================================
# Part 2: 数据清洗与处理
# ============================================
# 检查缺失值
print("\n缺失值统计：")
print(combined.isnull().sum())

# 填充缺失值（用前一日数据填充）
combined = combined.ffill()

# 添加年份和月份列用于分组分析
combined['Year'] = combined.index.year
combined['Month'] = combined.index.month

# ============================================
# Part 3: 可视化分析
# ============================================
# 设置图形风格
sns.set_style("whitegrid")
sns.set_palette("husl")

# ----- 图1:完整历史趋势图 -----
fig, ax = plt.subplots(figsize=(14, 7))
ax.plot(combined.index, combined['USD/CNY'], label='美元兑人民币 (USD/CNY)', linewidth=1.5)
ax.plot(combined.index, combined['EUR/CNY'], label='欧元兑人民币 (EUR/CNY)', linewidth=1.5)
ax.set_title('人民币兑美元/欧元历史汇率走势 (2015-2026)', fontsize=16, fontweight='bold')
ax.set_xlabel('日期', fontsize=12)
ax.set_ylabel('汇率 (人民币/外币)', fontsize=12)
ax.legend(fontsize=12)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('figures/exchange_rate_trend.png', dpi=150)
plt.show()


# ----- 图2:近5年走势更清晰-----
recent = combined.loc['2021-01-01':]
fig, ax = plt.subplots(figsize=(14, 6))
ax.plot(recent.index, recent['USD/CNY'], label='USD/CNY', linewidth=2, color='#2E86AB')
ax.plot(recent.index, recent['EUR/CNY'], label='EUR/CNY', linewidth=2, color='#A23B72')
ax.set_title('人民币兑美元/欧元汇率走势 (2021-2026)', fontsize=14, fontweight='bold')
ax.set_xlabel('日期', fontsize=12)
ax.set_ylabel('汇率', fontsize=12)
ax.legend(fontsize=12)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('figures/recent_5years.png', dpi=150)
plt.show()


# ----- 图3:年度平均汇率对比（柱状图）-----
yearly_avg = combined.groupby('Year')[['USD/CNY', 'EUR/CNY']].mean()
fig, ax = plt.subplots(figsize=(14, 6))
x = range(len(yearly_avg.index))
width = 0.35
ax.bar([i - width/2 for i in x], yearly_avg['USD/CNY'], width, label='USD/CNY', color='#2E86AB')
ax.bar([i + width/2 for i in x], yearly_avg['EUR/CNY'], width, label='EUR/CNY', color='#A23B72')
ax.set_xticks(x)
ax.set_xticklabels(yearly_avg.index, rotation=45)
ax.set_title('人民币兑美元/欧元年度平均汇率 (2015-2025)', fontsize=14, fontweight='bold')
ax.set_xlabel('年份', fontsize=12)
ax.set_ylabel('平均汇率', fontsize=12)
ax.legend(fontsize=12)
ax.grid(True, axis='y', alpha=0.3)
plt.tight_layout()
plt.savefig('figures/yearly_avg.png', dpi=150)
plt.show()


# ============================================
# Part 4: 生成分析结论（打印到输出）
# ============================================
print("\n" + "="*50)
print("分析结论")
print("="*50)

# 最新汇率
latest_usd = combined['USD/CNY'].iloc[-1]
latest_eur = combined['EUR/CNY'].iloc[-1]
print(f"1. 截至最新数据日,1美元 ≈ {latest_usd:.4f} 人民币:1欧元 ≈ {latest_eur:.4f} 人民币。")

# 汇率波动幅度
usd_range = combined['USD/CNY'].max() - combined['USD/CNY'].min()
eur_range = combined['EUR/CNY'].max() - combined['EUR/CNY'].min()
print(f"2. 历史最高值:USD/CNY最高 {combined['USD/CNY'].max():.4f},EUR/CNY最高 {combined['EUR/CNY'].max():.4f}")
print(f"3. 历史最低值:USD/CNY最低 {combined['USD/CNY'].min():.4f},EUR/CNY最低 {combined['EUR/CNY'].min():.4f}")

# 近5年趋势
if pd.Timestamp('2021-01-01') in combined.index:
    recent_usd_start = combined.loc['2021-01-01', 'USD/CNY']
    recent_usd_end = combined['USD/CNY'].iloc[-1]
    recent_eur_start = combined.loc['2021-01-01', 'EUR/CNY']
    recent_eur_end = combined['EUR/CNY'].iloc[-1]
    usd_change = (recent_usd_end - recent_usd_start) / recent_usd_start * 100
    eur_change = (recent_eur_end - recent_eur_start) / recent_eur_start * 100
    print(f"4. 近5年变化:美元兑人民币 {usd_change:+.2f}%，欧元兑人民币 {eur_change:+.2f}%")
else:
    print("4. 数据不足5年，无法计算近5年变化。")

print("\n分析完成！图表已保存到figures目录。")
```

## 分析结论

根据获取到的历史汇率数据（时间范围约为 2022–2026 年），对人民币兑美元、欧元的走势分析如下：

### 1. 最新汇率水平
截至最新数据日，1 美元 ≈ 7.1771 人民币，1 欧元 ≈ 7.6917 人民币。当前欧元兑人民币汇率高于美元，与历史经验一致，反映欧元区整体物价和币值水平。

### 2. 历史波动区间
- **美元兑人民币**：在 6.8374 至 7.2258 之间波动，振幅约 0.3884，相对稳定。
- **欧元兑人民币**：在 7.3194 至 8.0879 之间波动，振幅约 0.7685，**波动幅度明显大于美元**，与欧元在国际外汇市场的高波动特性相符。

### 3. 汇率走势解读
- 美元兑人民币接近历史高点（7.2258），表明近期人民币对美元存在一定贬值压力。这主要受美联储维持高利率、中美利差倒挂以及地缘政治因素影响。
- 欧元兑人民币虽从高点 8.0879 回落至 7.69 附近，但仍处于历史中等偏上位置。欧洲央行加息周期和欧元区经济疲软的博弈，导致欧元波动加剧。

### 4. 对用户的影响
- **留学/旅游**：当前美元兑人民币接近 7.18，赴美成本较高；欧元 7.69 亦不便宜，换汇可关注回调机会。
- **进出口企业**：人民币对美元贬值利好出口企业（以美元结算），但进口原材料成本上升。欧元波动大，需要做好汇率风险管理。

### 5. 数据局限性
由于本次获取的数据时间跨度不足 5 年，无法分析长期趋势和 2015 年汇改等历史事件的影响。建议后续补充更长时间范围的数据，以获得更全面的结论。
